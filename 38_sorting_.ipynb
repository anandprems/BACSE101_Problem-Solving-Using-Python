```python
numbers = [5, 3, 8, 4, 2]

for i in range(len(numbers)):
    swapped = False
    for j in range(len(numbers) - 1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
            swapped = True
    if not swapped:   
        break

print("Sorted List:", numbers)
```

```python
numbers = [5, 3, 8, 4, 2]

n = len(numbers)
for i in range(n):
    for j in range(n - i - 1): 
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print("Sorted List:", numbers)
```

```python
numbers = [29, 10, 14, 37, 13]

for i in range(len(numbers)):
    min_index = i
    for j in range(i+1, len(numbers)):
        if numbers[j] < numbers[min_index]:
            min_index = j
    # Swap
    numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

print("Sorted list:", numbers)
```

```python
numbers = [12, 11, 13, 5, 6]

for i in range(1, len(numbers)):  
    key = numbers[i]
    j = i - 1
    while j >= 0 and numbers[j] > key:   
        numbers[j + 1] = numbers[j]
        j -= 1
    numbers[j + 1] = key  

print("Sorted list:", numbers)
```

```python
arr = [12, 4, 7, 9, 1, 15]

pivot = arr[0]
left = []
right = []

for x in arr[1:]:
    if x < pivot:
        left.append(x)
    else:
        right.append(x)

print("Step 1:", left, [pivot], right)

pivot2 = left[0]
left2 = []
right2 = []

for x in left[1:]:
    if x < pivot2:
        left2.append(x)
    else:
        right2.append(x)

print("Step 2:", left2, [pivot2], right2)
```

```python
# Initial array
arr = [12, 4, 7, 9, 1, 15]

pivot1 = arr[0]
left1 = []
right1 = []

for x in arr[1:]:
    if x < pivot1:
        left1.append(x)
    else:
        right1.append(x)

print("Step 1:", left1, [pivot1], right1)


pivot2 = left1[0]
left2 = []
right2 = []

for x in left1[1:]:
    if x < pivot2:
        left2.append(x)
    else:
        right2.append(x)

print("Step 2:", left2, [pivot2], right2)


pivot3 = right2[0]
left3 = []
right3 = []

for x in right2[1:]:
    if x < pivot3:
        left3.append(x)
    else:
        right3.append(x)

print("Step 3:", left3, [pivot3], right3)


final_sorted = left2 + [pivot2] + left3 + [pivot3] + right3 + [pivot1] + right1
print("Final Sorted:", final_sorted)
```

```python
arr = [12, 4, 7, 9, 1, 15, 20, 3]

stack = [arr]       
sorted_parts = []   

while stack:
    current = stack.pop()   

    if len(current) <= 1:
        sorted_parts.append(current)
        continue

    pivot = current[0]
    left = []
    right = []

    for x in current[1:]:
        if x < pivot:
            left.append(x)
        else:
            right.append(x)

    if right:
        stack.append(right)
    stack.append([pivot])  
    if left:
        stack.append(left)

final_sorted = []
for part in sorted_parts[::-1]:   
    final_sorted.extend(part)

print("Final Sorted:", final_sorted)
```
