```python
from pygorithm.searching import linear_search
numbers = [10, 20, 30, 40, 50]
index = linear_search.search(numbers, 30)
print("Index of 30:", index)
print(linear_search.get_code())
print(linear_search.time_complexities())
```

```python
from pygorithm.searching import binary_search
numbers = [10, 20, 30, 40, 50]
index = binary_search.search(numbers, 40)
print("Index of 40:", index)
print(binary_search.get_code())
print(binary_search.time_complexities())
```

```python
import time
from pygorithm.searching import linear_search, binary_search
nums = list(range(1, 1000000))
start = time.time()
linear_search.search(nums, 999999)
end = time.time()
print("Linear Search Time:", end - start)
start = time.time()
binary_search.search(nums, 999999)
end = time.time()
print("Binary Search Time:", end - start)
```
